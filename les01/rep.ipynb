## Ответы на контрольные вопросы

### 1. Типы ячеек в Jupyter Notebook
- **Code** — исполняемый код. Результат выполнения может быть текстом, таблицей, изображением или другим выводом.
- **Markdown** — текст с разметкой: заголовки, списки, ссылки, изображения, формулы и пояснения.
- **Raw** — необрабатываемый текст, который не исполняется как код и может использоваться при экспорте документа.

В интерфейсе Google Colab основными являются ячейки кода и текстовые ячейки.

### 2. Как с помощью Markdown сформировать иерархический список?
Вложенные пункты записывают с отступом относительно родительского пункта. Например, можно использовать четыре пробела:

```markdown
1. Первый этап
    - Подэтап А
    - Подэтап Б
2. Второй этап
```

### 3. Как с помощью Markdown оформить цитату?
В начале строки ставят символ `>` и пробел:

```markdown
> Текст цитаты.
```

Для вложенной цитаты используют `>>`.

### 4. Как в текстовую ячейку вставить отформатированный блок кода?
Блок окружают строками с тремя обратными кавычками. После открывающих кавычек можно указать язык для подсветки синтаксиса:

````markdown
```python
x = 2 + 3
print(x)
```
````

Такой блок служит для отображения кода. Для выполнения код помещают в ячейку Code.

### 5. Сколько уровней заголовка существует в Markdown?
Шесть: от `# Заголовок первого уровня` до `###### Заголовок шестого уровня`. После символов `#` ставится пробел.


### 6. Шаги процесса анализа данных
Процесс обычно включает следующие этапы; к предыдущим этапам можно возвращаться по мере появления новых результатов.

1. **Постановка задачи.** Определить вопрос, цель анализа, ограничения и критерии успешного результата.
2. **Сбор данных.** Получить данные из файлов, баз данных, опросов, наблюдений или API; проверить их происхождение и применимость к задаче.
3. **Очистка и подготовка.** Найти пропуски, дубликаты, ошибки и несогласованные форматы; привести данные к подходящему виду.
4. **Исследовательский анализ.** Рассчитать описательные статистики, построить графики, изучить распределения и связи между признаками.
5. **Проверка гипотез и моделирование, если нужны для задачи.** Выбрать методы, проверить предположения и оценить качество результата. При машинном обучении разделить данные для обучения и независимой проверки.
6. **Интерпретация и представление.** Ответить на исходный вопрос, объяснить результаты, ограничения и возможные действия.

**Подробнее: очистка и подготовка данных.** Сначала проверяют структуру таблицы, типы столбцов и допустимые значения. Затем выявляют пропуски и выясняют причины их возникновения. Например, пропущенную оценку нельзя автоматически заменять нулём: это изменит смысл данных и среднее значение. В зависимости от задачи пропуски оставляют, заполняют обоснованным способом или исключают отдельные записи. Дубликаты удаляют только после проверки, что это действительно повтор одной записи. Даты и единицы измерения приводят к единому формату. Выбросы исследуют: необычное значение может быть как ошибкой, так и реальным наблюдением. Все преобразования документируют, сохраняя исходные данные.

### 7. Типы данных
В анализе данных признаки часто делят на следующие типы:

- **Категориальные (качественные):**
    - *Номинальные* — категории без естественного порядка, например цвет или город.
    - *Порядковые* — категории с порядком, но без гарантированно равных расстояний, например «низкий», «средний», «высокий».
- **Числовые (количественные):**
    - *Дискретные* — отдельные счётные значения, например количество студентов.
    - *Непрерывные* — величины, которые могут принимать значения в некотором интервале, например масса или температура; на практике измерения имеют ограниченную точность.

**Подробнее: номинальные данные.** Они описывают принадлежность к категории. Для них можно считать частоты, доли и моду, строить столбчатые диаграммы. Среднее арифметическое кодов категорий не имеет содержательного смысла: если городам присвоить коды 1, 2 и 3, код 3 не означает «больше города», а средний код не описывает средний город.

Также различают шкалы измерения: номинальную, порядковую, интервальную и шкалу отношений. Для интервальной шкалы осмысленны разности (температура в °C), а для шкалы отношений есть содержательный абсолютный ноль и осмысленны отношения величин (масса).

Если вопрос касается **типов Python**, основные примеры: `int` (целые числа), `float` (числа с плавающей точкой), `bool` (логические значения), `str` (строки), `list` (списки), `tuple` (кортежи), `dict` (словари), `set` (множества). `None` обозначает отсутствие значения. Тип хранения и статистический тип признака могут различаться: числовой идентификатор студента является меткой, а не количественным признаком.


### 8. Что такое Jupyter Notebook, для чего используется?
Jupyter Notebook — среда для работы с интерактивными документами, объединяющими код, текст, формулы и результаты вычислений. Документ обычно сохраняется в формате `.ipynb`. Код выполняет вычислительное ядро; существуют ядра для разных языков.

Ноутбуки используют для анализа и визуализации данных, экспериментов, обучения и подготовки воспроизводимых отчётов. Поскольку ячейки можно выполнять в произвольном порядке, для проверки воспроизводимости полезно перезапустить ядро и выполнить все ячейки сверху вниз.

### 9. Почему Python популярен для анализа данных?
- Понятный синтаксис облегчает чтение кода и создание прототипов.
- Есть развитые библиотеки: NumPy для массивов и численных вычислений, pandas для таблиц, Matplotlib для графиков, SciPy для научных методов, scikit-learn для машинного обучения.
- Большое сообщество предоставляет документацию и учебные материалы.
- Python позволяет объединять загрузку, очистку, анализ и визуализацию данных в одном процессе.
- Многие ресурсоёмкие операции библиотек выполняются оптимизированным скомпилированным кодом. Популярность Python не означает, что любые циклы на чистом Python работают быстро.

### 10. Обязательно ли использовать Jupyter Notebook для анализа данных?
Нет. Анализ можно выполнять в обычных Python-скриптах, IDE, терминале, а также другими средствами, например R, SQL или электронными таблицами. Ноутбук удобен для исследования и пояснения результатов, а скрипты — для автоматизации и регулярного запуска обработки. Выбор зависит от задачи.
